```python
import pandas as pd
```

```python
df_plataforma = pd.read_csv("estudiantes_cursos.csv")
df_plataforma
```

```python
# Seleccionar una columna como Serie
puntajes = df_plataforma["Puntaje_Evaluacion"]

# Subconjunto bidimensional 
df_subconjunto = df_plataforma[
    ["Curso", "Modalidad", "Lecciones_Completadas"]
]

print(puntajes)
df_subconjunto
```

```python
fila = df_plataforma.loc[120]
fila
```

```python
# Eliminar todos los registros duplicados

df_duplicados = df_plataforma.drop_duplicates()
df_duplicados
```

```python
# Filtrar estudiantes con más de 40 lecciones completadas

df_plataforma[df_plataforma["Lecciones_Completadas"] > 40]
```

```python
#Registros de estudiantes inscritos 
df_plataforma[df_plataforma["Curso"].isin(["Python para Datos", "Arquitectura Cloud"])]
```

```python
# Filtrar alumnos con puntajes entre 70.0 y 85.0

df_plataforma[
    df_plataforma["Puntaje_Evaluacion"].between(70.0, 85.0)
]
```

```python
# Filtrar por modalidad y horas de estudio

df_plataforma.query('Modalidad == "Asincrónico" and Horas_Estudio > 25')
```

```python
# Filtrar por nivel educativo y nombre del curso

df_plataforma[
    (df_plataforma["Nivel_Educativo"] == "Profesional") &
    (df_plataforma["Curso"].str.startswith("B"))
]
```

```python
# Reemplazar los valores nulos de Horas_Estudio por el promedio
df_plataforma["Horas_Estudio"] = df_plataforma["Horas_Estudio"].fillna(
    df_plataforma["Horas_Estudio"].mean()
)

# Reemplazar los valores nulos de Puntaje_Evaluacion por 0.0
df_plataforma["Puntaje_Evaluacion"] = df_plataforma["Puntaje_Evaluacion"].fillna(0.0)

df_plataforma
```

```python

def evaluar_nota(nota):
    return "Aprobado" if nota >= 70.0 else "Reprobado"


df_plataforma["Estado_Materia"] = df_plataforma["Puntaje_Evaluacion"].apply(evaluar_nota)


df_plataforma[["Puntaje_Evaluacion", "Estado_Materia"]]
```

```python
# Renombrar columnas
df_plataforma = df_plataforma.rename(columns={
    "Modalidad": "Formato_Clase",
    "Certificado_Generado": "Graduado"
})

# Ordenar columnas
columnas_ordenadas = ["Estado_Materia", "ID_Estudiante"] + [
    columna for columna in df_plataforma.columns
    if columna not in ["Estado_Materia", "ID_Estudiante"]
]

df_plataforma = df_plataforma[columnas_ordenadas]

df_plataforma
```

```python
#Agrupar y calcular el promedio 

df_plataforma.groupby("Curso").agg({
    "Horas_Estudio": "mean",
    "Lecciones_Completadas": "sum"
})
```

```python
def rango(series):
    return series.max() - series.min()

df_plataforma.groupby("Nivel_Educativo").agg({
    "Puntaje_Evaluacion": rango
})
```

```python
df_plataforma.groupby(["Curso", "Formato_Clase"]).agg({
    "Horas_Estudio": "sum",
    "Puntaje_Evaluacion": "mean"
})
```
